```python
# Code to order tasks
def ordertasks(files):
  [fileA, fileB] = files
  sectionA = fileA.split("-")[0]
  sectionB = fileB.split("-")[0]
  if sectionA == sectionB:
    # It's down to task number
    taskA = fileA.split("-")[1]
    taskB = fileB.split("-")[1]
    return [fileA, fileB] if taskA < taskB else [fileB, fileA]
  else:
    match sectionA:
      case "Introduction":
        # Then task A is definitely before task B
        return [fileA, fileB]
      case "Conclusion":
        # Then task B is definitely before task A
        return [fileB, fileA]
      case _:
        match sectionB:
          case "Introduction":
            return [fileB, fileA]
          case "Conclusion":
            return [fileA, fileB]
          case _:
            sectionA_no=sectionA.split()[1]
            sectionB_no=sectionB.split()[1]
            return [fileA, fileB] if sectionA_no < sectionB_no else [fileB, fileA]
```

```python
def findSectionAndTask(f):
  section = f.split("-")[1]
  task = f.split("-")[2].split(".xlsx")[0]
  return '{}-{}'.format(section, task)
```
